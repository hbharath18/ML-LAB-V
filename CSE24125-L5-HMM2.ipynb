```python
# Install once
# pip install hmmlearn

import numpy as np
from hmmlearn.hmm import CategoricalHMM

# -----------------------------
# Hidden States
# -----------------------------
states = ["Walking", "Running", "Resting"]

# -----------------------------
# HMM Model
# -----------------------------
model = CategoricalHMM(
    n_components=3,
    init_params="",      # Do not overwrite our parameters
    params="ste"         # Parameters to learn during training
)

# Initial State Probability
model.startprob_ = np.array([
    0.5,
    0.3,
    0.2
])

# Transition Matrix
model.transmat_ = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
])

# Emission Matrix
model.emissionprob_ = np.array([
    [0.6, 0.3, 0.1],   # Walking
    [0.1, 0.3, 0.6],   # Running
    [0.7, 0.2, 0.1]    # Resting
])

# -----------------------------
# Observation Sequences
# Low=0 Medium=1 High=2
# -----------------------------
O  = np.array([[2],[1],[0],[2]])
O1 = np.array([[0],[1],[2],[1]])
O2 = np.array([[2],[2],[2],[2]])
O3 = np.array([[0],[0],[0],[0]])
O4 = np.array([[1],[1],[1],[1]])

# -----------------------------
# Task 3 & 4
# Likelihood
# -----------------------------
print("Likelihood of O  =", np.exp(model.score(O)))
print("Likelihood of O1 =", np.exp(model.score(O1)))
print("Likelihood of O2 =", np.exp(model.score(O2)))
print("Likelihood of O3 =", np.exp(model.score(O3)))
print("Likelihood of O4 =", np.exp(model.score(O4)))

# -----------------------------
# Task 6
# Most Probable Hidden State Sequence
# -----------------------------
logprob, sequence = model.decode(O, algorithm="viterbi")

print("\nMost Probable Hidden State Sequence:")

for i in sequence:
    print(states[i], end=" ")

print("\n")

# -----------------------------
# Task 7
# Baum-Welch Training
# -----------------------------
model.fit(O)

# -----------------------------
# Task 8
# Learned Parameters
# -----------------------------
print("Learned Initial State Probability")
print(model.startprob_)

print("\nLearned Transition Matrix")
print(model.transmat_)

print("\nLearned Emission Matrix")
print(model.emissionprob_)

# -----------------------------
# Comparison
# -----------------------------
print("\nOriginal Initial State Probability")
print(np.array([0.5,0.3,0.2]))

print("\nOriginal Transition Matrix")
print(np.array([
    [0.6,0.3,0.1],
    [0.2,0.5,0.3],
    [0.3,0.2,0.5]
]))

print("\nOriginal Emission Matrix")
print(np.array([
    [0.6,0.3,0.1],
    [0.1,0.3,0.6],
    [0.7,0.2,0.1]
]))
```
