```python
#Importing Libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
#New unknown libraries
from sklearn.linear_model import LinearRegression
                                        #for Linear Regression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score 
                                        #for question 9 calculations
from sklearn.model_selection import train_test_split
                                        #for qn 3



#1.Create the dataset using Pandas and display the first few records
data={
    "House_Size":[600,800,1000,1200,1400,1600,1800,2000,2200,2400],
    "House_Prize":[25,32,38,45,52,58,65,72,80,87]
}
df=pd.DataFrame(data)
print("---First Few Records---")
print(df.head(),"\n")



#2.Perform exploratory data visualization by plotting a scatter plot between House Size and House Price. Briefly comment on the relationship between the variables.
plt.figure(figsize=(6,4))
plt.scatter(df["House_Size"],df["House_Prize"], color="green")
plt.title("House Size vs House Prize")
plt.xlabel("House Size (sq.ft.)")
plt.ylabel("House Price (Lakhs)")
plt.grid(True)
plt.show()
print("Here, we can observe Strong Linear Relationship between House Size & House Prize")



#3.Split the dataset into training and testing sets using an 80:20 ratio.
X=df[["House_Size"]]
Y=df["House_Prize"]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2,random_state=42)



#4. Build a Simple Linear Regression model using the training data.
model=LinearRegression()
model.fit(X_train, Y_train)



#5. Display the learned model parameters: Regression Constant (θ0) & Regression Coefficient (θ1)
print(f"Regression Constant (Intercept,theta0):{model.intercept_:.4f}")
print(f"Regression Coefficient (Slope,theta1):{model.coef_[0]:.4f}\n")


#6. Predict the house prices for the test dataset and compare the predicted values with the actual values.

```
